```python
print("=== KASIR PINTAR KANTIN SMK RPL ===")
nama_barang = input("Nama Makanan/Minuman : ")
harga = int(input("Harga Satuan (Rp)    : "))
jumlah = int(input("Jumlah Beli          : "))

total_belanja = harga * jumlah
```

```python
# 1. Cek Diskon Belanja
if total_belanja >= 60000:
    diskon = 5000
    print("Selamat! Anda mendapat diskon Rp5.000")
else:
    diskon = 0

bayar_akhir = total_belanja - diskon
print(f"Total yang harus dibayar: Rp{bayar_akhir}")
print("-----------------------------------")
```

```python
# 2. Input Pembayaran & Validasi Uang Cukup/Kurang
uang_dibayar = int(input("Masukkan Uang Tunai Pembeli (Rp): "))

if uang_dibayar < bayar_akhir:
    uang_kurang = bayar_akhir - uang_dibayar
    print(f"Maaf, uang Anda kurang Rp{uang_kurang}! Transaksi gagal.")
else:
    kembalian = uang_dibayar - bayar_akhir
    print(f"Transaksi Berhasil! Uang Kembalian Anda: Rp{kembalian}")
```
