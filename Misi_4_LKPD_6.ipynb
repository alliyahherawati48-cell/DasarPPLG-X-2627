```python
nama = input("Masukkan Nama Siswa: ")
nilai_ujian = int(input("Masukkan Nilai Ujian (0-100): "))
kehadiran = int(input("Masukkan Persentase Kehadiran (0-100): "))
```

```python
status_lulus = (nilai_ujian >= 75) and (kehadiran >= 80)
```

```python
print("------------------------------------------------")
print(f"Nama Siswa          : {nama}")
print(f"Apakah Lulus Syarat?: {status_lulus}")
print("------------------------------------------------")
```
