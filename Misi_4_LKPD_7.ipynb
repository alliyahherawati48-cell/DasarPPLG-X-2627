```python
print("=== CEK SYARAT TIKET ROLLER COASTER ===")
nama = input("Nama Pengunjung        : ")
umur = int(input("Masukkan Umur (tahun)  : "))
```

```python
# Pengecekan lapis pertama: Umur minimal 12 tahun
if umur >= 12:
    tinggi = int(input("Masukkan Tinggi Badan (cm): "))
 # Pengecekan lapis kedua (Nested If): Tinggi badan minimal 145 cm
    if tinggi >= 145:
        print(f"Boleh Naik Wahana! Selamat menikmati, {nama}.")
    else:
        print(f"Maaf, umur cukup tapi tinggi badan belum memenuhi syarat, {nama}.")
else:
    print(f"Maaf, belum cukup umur! ({nama} berusia {umur} tahun)")
```
